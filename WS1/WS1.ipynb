## First LLM demo

We will now load a lightweight language model and test a simple prompt.

This is only a first demonstration. At this stage, the model is not yet connected to a retrieval system. We are simply exploring how an LLM responds to instructions.

Later in the workshop series, we will improve this by giving the model relevant context from documents.

## The language model used in this workshop

In this workshop, we use a lightweight language model called:

**Qwen2.5-0.5B-Instruct**

### What is this model?

The Qwen2.5-0.5B-Instruct model is developed by **Alibaba Cloud** as part of the **Qwen (Tongyi Qianwen)** family of language models.

The Qwen models are a series of open large language models designed for research and practical applications, including text generation, reasoning, and instruction-following tasks.

### Why do we use this model?

We selected this model for several practical reasons:

- **Lightweight and fast**  
  It is small enough to run smoothly in Google Colab without requiring powerful hardware.

- **No access restrictions**  
  The model is openly available (not gated), so you can use it without requesting permissions.

- **Good for learning**  
  It is capable enough to demonstrate key concepts such as prompting and RAG, while still showing some limitations.

### Important note

Because this is a small model, its answers are not always accurate.  
It may sometimes generate incorrect or misleading information.  

This is expected, and it helps us understand why techniques like **Retrieval-Augmented Generation (RAG)** are important.

### Learn more

You can explore the model here:

- https://huggingface.co/Qwen/Qwen2.5-0.5B-Instruct

```python
# Load a lightweight instruction-following model.
# We keep the setup simple for Google Colab.
from transformers import pipeline

model_name = "Qwen/Qwen2.5-0.5B-Instruct"

generator = pipeline(
    "text-generation",
    model=model_name,
    tokenizer=model_name
)

print("Model loaded successfully.")
```

## Try a simple prompt

In the next cell, we ask the model a general question.

This helps us see how an LLM responds using its own knowledge, without any external documents.

```python
prompt = "Explain what a healthy lifestyle is in simple terms."

response = generator(
    prompt,
    max_new_tokens=120
)
print(response[0]["generated_text"])
```

```python
#Let's try another prompt
prompt = "What are the top 10 best attractions to visit in Bristol (England) 2026?"

response = generator(
    prompt,
    max_new_tokens=250
)

print(response[0]["generated_text"])
```

The model's answer is not fully correct.
This is expected, especially with smaller models.

Language models can produce answers that sound convincing
but are not always accurate.

This is one of the key reasons why we use RAG:
to ground answers in real, trusted documents.

## Key takeaways

- Large Language Models (LLMs) can generate useful responses from prompts
- However, they do not always produce accurate or reliable information
- They may generate answers that sound correct but are not grounded in real data
- This limitation is known as hallucination

To improve reliability, we need to connect the model to real documents.

This leads to the next step: **retrieval**.
